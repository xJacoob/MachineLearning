```python
from pypdf import PdfReader
from pathlib import Path
import numpy as np
import ollama

DATA_DIR = Path.cwd() / "data"

files = {}

for file in DATA_DIR.glob("*.pdf"):
    reader = PdfReader(file)
    name = file.name.replace(".pdf", "")
    files[name] = reader

files
```

```python
schedule = files["harmonogram"]
isi = files["isi"]
statue = files["regulamin"]
scholarship = files["stypendia"]

print(len(schedule.pages))
print(len(isi.pages))
print(len(statue.pages))
print(len(scholarship.pages))
```
